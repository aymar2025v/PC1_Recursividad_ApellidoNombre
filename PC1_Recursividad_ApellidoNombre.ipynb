## Actividad 1: Suma de dígitos recursiva y traza de pila

Implementar `suma_digitos(n)` que sume los dígitos de un entero positivo.
- Caso base: si `n` tiene un solo dígito, retornar `n`.
- Caso recursivo: sumar el último dígito con `suma_digitos(n // 10)`.

````python
def suma_digitos(n: int) -> int:
    """Suma los dígitos de n usando recursividad."""
    if n < 10:
        return n
    return n % 10 + suma_digitos(n // 10)
````

````python
assert suma_digitos(12345) == 15, 'Error: 1+2+3+4+5=15'
assert suma_digitos(0) == 0, 'Error: caso base n=0'
assert suma_digitos(9) == 9, 'Error: un solo dígito'

print('suma_digitos(12345) =', suma_digitos(12345))
print('suma_digitos(0)     =', suma_digitos(0))
print('suma_digitos(9)     =', suma_digitos(9))
````

## Actividad 2: Búsqueda Binaria con conteo de comparaciones

Implementar `busqueda_binaria(arr, objetivo)` que retorne `(índice, comparaciones)`.
Si no se encuentra, retornar `(-1, comparaciones)`.

````python
def busqueda_binaria(arr: List[int], objetivo: int) -> Tuple[int, int]:
    """Retorna (índice, n comparaciones). -1 si no encontrado."""
    bajo, alto, comparaciones = 0, len(arr) - 1, 0
    while bajo <= alto:
        comparaciones += 1
        medio = (bajo + alto) // 2
        if arr[medio] == objetivo:
            return medio, comparaciones
        elif arr[medio] < objetivo:
            bajo = medio + 1
        else:
            alto = medio - 1
    return -1, comparaciones
````

````python
arr = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
idx, comp = busqueda_binaria(arr, 23)
print(f'23 encontrado en índice {idx}, comparaciones={comp}')

cota_teorica = math.ceil(math.log2(len(arr) + 1))
print(f'Cota teórica: ceil(log2({len(arr)}+1)) = {cota_teorica} comparaciones')
assert comp <= cota_teorica, f'Supera la cota teórica: {comp} > {cota_teorica}'
print('Cota verificada ✓')
````

### SALIDA ESPERADA
```bash
23 encontrado en índice 5, comparaciones=3
Cota teórica: ceil(log2(10+1)) = 4 comparaciones
Cota verificada ✓
```

````python
def max_comparaciones_binaria(n: int) -> int:
    """Mide el máximo número de comparaciones para un arreglo ordenado de tamaño n."""
    arr = list(range(n))
    max_comp = 0
    for objetivo in arr:
        _, comp = busqueda_binaria(arr, objetivo)
        if comp > max_comp:
            max_comp = comp
    return max_comp

def max_comparaciones_lineal(n: int) -> int:
    """Peor caso búsqueda lineal: n comparaciones."""
    return n

valores_n = [10, 100, 1_000, 10_000, 100_000, 1_000_000]
resultados = []
for n in valores_n:
    lineal = max_comparaciones_lineal(n)
    binaria = max_comparaciones_binaria(n)
    cota = math.ceil(math.log2(n + 1))
    cumple = binaria <= cota
    resultados.append({
        'n': n,
        'Lineal (max comp.)': lineal,
        'Binaria (max comp.)': binaria,
        'Cota teórica ⌈log2(n+1)⌉': cota,
        '¿Cota cumplida?': 'Sí' if cumple else 'No'
    })

df = pd.DataFrame(resultados)
df
````

## Actividad 3: Merge Sort — implementación completa y benchmark

Implementar `merge_sort` y `merge`, ordenar códigos de matrícula y medir tiempos.

````python
def merge_sort(arr: List[int]) -> List[int]:
    if len(arr) <= 1:
        return arr[:]
    mid = len(arr) // 2
    izq = merge_sort(arr[:mid])
    der = merge_sort(arr[mid:])
    return merge(izq, der)

def merge(izq: List[int], der: List[int]) -> List[int]:
    res = []
    i = j = 0
    while i < len(izq) and j < len(der):
        if izq[i] <= der[j]:
            res.append(izq[i])
            i += 1
        else:
            res.append(der[j])
            j += 1
    res.extend(izq[i:])
    res.extend(der[j:])
    return res
````

````python
codigos_originales = [
    20211200, 20210300, 20211050, 20210780, 20211500,
    20210100, 20211800, 20210640, 20210950, 20211400
]

t0 = time.perf_counter()
codigos_ordenados = merge_sort(codigos_originales)
ms = (time.perf_counter() - t0) * 1000

print('Ordenados:', codigos_ordenados)
assert codigos_ordenados == sorted(codigos_originales), 'Error en Merge Sort'
print(f'Tiempo (n=10): {ms:.4f} ms')
````

### SALIDA ESPERADA


````python
for n in [100, 1_000, 10_000, 100_000]:
    datos = random.sample(range(20_000_000, 29_999_999), n)
    t0 = time.perf_counter()
    merge_sort(datos)
    ms_n = (time.perf_counter() - t0) * 1000
    print(f'n={n:>8}  tiempo={ms_n:10.3f} ms')
````

## Actividad 4: Teorema Maestro — resolución de recurrencias

Para cada recurrencia, identificar `a`, `b`, `f(n)`, calcular `log_b(a)`, determinar el caso y enunciar la solución Θ(·).

### (a) T(n) = 4T(n/2) + O(n)
- a = 4, b = 2, f(n) = O(n)
- log₂(4) = 2
- f(n) = O(n) = O(n^(2-1)) → Caso 1
- **Solución: Θ(n²)**

### (b) T(n) = T(n/3) + O(1)
- a = 1, b = 3, f(n) = O(1)
- log₃(1) = 0
- f(n) = Θ(n⁰) → Caso 2
- **Solución: Θ(log n)**

### (c) T(n) = 2T(n/2) + O(n²)
- a = 2, b = 2, f(n) = O(n²)
- log₂(2) = 1
- f(n) = Ω(n^(1+1)) → Caso 3 (verifica regularidad: 2·(n/2)² = n²/2 ≤ c·n², c = 1/2)
- **Solución: Θ(n²)**

### (d) T(n) = 3T(n/4) + O(n)
- a = 3, b = 4, f(n) = O(n)
- log₄(3) ≈ 0.792
- f(n) = Ω(n^(0.792+ε)) → Caso 3 (verifica regularidad: 3·(n/4) = 3n/4 ≤ c·n, c = 3/4)
- **Solución: Θ(n)**

## Preguntas de análisis

1. **¿Qué debe retornar `suma_digitos(0)`?**
   Retorna 0. En la implementación, `n < 10` cubre el caso base y devuelve `n`, por lo que `0` se maneja correctamente.

2. **¿Por qué la Búsqueda Binaria requiere arreglo ordenado?**
   Porque descarta mitades basándose en comparaciones. Ejemplo con `arr = [3, 1, 2]` y objetivo `3`:
   - bajo=0, alto=2, medio=1 → arr[1]=1 < 3 → bajo=2
   - medio=2, arr[2]=2 < 3 → bajo=3 → termina
   - Retorna -1 (falso negativo), aunque `3` está en el índice 0.

3. **Recurrencia (c) T(n) = 2T(n/2) + O(n²): ¿qué algoritmo real podría tenerla?**
   Un algoritmo que divide el problema en 2 subproblemas de tamaño n/2 y luego realiza un trabajo local cuadrático para combinarlos. No es preferible a Merge Sort porque O(n²) es peor que O(n log n) para n grandes.

4. **Peor y mejor caso de Merge Sort.**
   Peor caso: O(n log n) comparaciones. Mejor caso: también O(n log n). A diferencia de Quick Sort, Merge Sort siempre realiza aproximadamente el mismo número de operaciones para un `n` dado, independientemente de la entrada.
